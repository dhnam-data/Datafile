귀무가설(H0): 심화교육 전 시험 점수와 심화교육 후 시험 점수의 평균값은 같다.
- (심화교육 전후 통계학 시험 점수의 차이에 대한 모집단 평균은 0이다. [H0: μd ＝ 0])


대립가설(H1): 심화교육 전 시험 점수와 심화교육 후 시험 점수의 평균값은 같지 않다.
- (심화교육 전후 통계학 시험 점수의 차이에 대한 모집단 평균은 0이 아니다.[H1: μd ≠ 0])

※ **μd = μ1 - μ2**

```python
import pandas as pd
pd.set_option('display.max_columns', None) # 컬럼 전체보기

df = pd.read_excel("https://github.com/dhnam-data/Datafile/raw/refs/heads/main/StatisticsHealthCare/data/13-1.%ED%8F%89%EA%B7%A0%EB%B9%84%EA%B5%90.xlsx")

print(df.shape)
df.head()
```

```python
# pre: 교육전
# post: 교육후

# 결측치 제거
df = df.dropna(subset=['pre', 'post'])

from scipy import stats

display(stats.ttest_rel(df['pre'], df['post']))
```
