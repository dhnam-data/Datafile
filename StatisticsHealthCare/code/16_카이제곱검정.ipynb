귀무가설(H0): 성별과 1년간 입원여부는 서로 독립적이다.

대립가설(H1): 성별과 1년간 입원여부는 서로 독립적이지 않다.

```python
import pandas as pd
pd.set_option('display.max_columns', None) # 컬럼 전체보기

# 파일 불러오기
df = pd.read_excel("https://github.com/dhnam-data/Datafile/raw/refs/heads/main/StatisticsHealthCare/data/16.%EC%B9%B4%EC%9D%B4%EC%A0%9C%EA%B3%B1%EA%B2%80%EC%A0%95.xlsx")
print(df.shape)
df.head()
```

```python
# h20_g3: 성별
# adm: 1년간 입원여부
df = df[['h20_g3', 'adm']].dropna()

# 교차표 생성
table = pd.crosstab(df['h20_g3'], df['adm'])


from scipy import stats

result = stats.chi2_contingency(table, correction=False)

print(result[0],'\n') # 카이제곱 통계량
print(result[1],'\n') # p값
print(result[2],'\n') # 자유도
print(result[3],'\n') # 기대빈도
print(table) # 관측빈도

```
