귀무가설(H0): 일반가구와 저소득층 가구의 평균 민간의료보험료는 같다. [H0: μ1 ＝ μ2]

대립가설(H1): 일반가구와 저소득층 가구의 평균 민간의료보험료는 같지 않다. [H1: μ1 ≠ μ2]

```python
import pandas as pd
pd.set_option('display.max_columns', None) # 컬럼 전체보기


# 파일 불러오기
df = pd.read_excel("https://github.com/dhnam-data/Datafile/raw/refs/heads/main/StatisticsHealthCare/data/1.%ED%8F%89%EA%B7%A0%EB%B9%84%EA%B5%90.xlsx")

print(df.shape)
df.head()
```

```python
df = df.dropna(subset=['h2005_3aq2'])
cond1 = df['h20_hc_n_all'] == 1 # 일반가구
cond2 = df['h20_hc_n_all'] == 2 # 저소득층 가구
gen_df = df[cond1]
low_df = df[cond2]
```

```python
from scipy import stats

levene_result = stats.levene(
    gen_df['h2005_3aq2'],
    low_df['h2005_3aq2'],
    center='mean'
)

print('Levene 통계량:', levene_result.statistic)
print('p값:', levene_result.pvalue)
```

```python
from scipy import stats

# 등분산
display(stats.ttest_ind(gen_df['h2005_3aq2'], low_df['h2005_3aq2'], equal_var=True))

# 등분산을 가정하지 않음
display(stats.ttest_ind(gen_df['h2005_3aq2'], low_df['h2005_3aq2'], equal_var=False))
```
