귀무가설(H0): 조사대상자의 경상소득은 민간의료보험 납입액에 영향을 미친다.

대립가설(H1): 조사대상자의 경상소득은 민간의료보험 납입액에 영향을 미치지 않는다.

```python
import pandas as pd
pd.set_option('display.max_columns', None) # 컬럼 전체보기

# 파일 불러오기
df = pd.read_excel("https://github.com/dhnam-data/Datafile/raw/refs/heads/main/StatisticsHealthCare/data/4.%ED%9A%8C%EA%B7%80%EB%B6%84%EC%84%9D.xlsx")
print(df.shape)
df.head()


```

```python
# h20_cin: 경상소득
# h2005_3aq2: 민간의료보험 월납부액
df = df[['h20_cin', 'h2005_3aq2']].dropna()

from statsmodels.formula.api import ols

result = ols('h2005_3aq2 ~ h20_cin', df).fit()

print(result.summary())
```
