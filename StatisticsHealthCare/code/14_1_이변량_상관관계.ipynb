귀무가설(H0): 경상소득과 민간의료보험료의 모집단 상관계수는 0이다.

대립가설(H1): 경상소득과 민간의료보험료의 모집단 상관계수는 0이 아니다.

```python
import pandas as pd
pd.set_option('display.max_columns', None) # 컬럼 전체보기

# 파일 불러오기
df = pd.read_excel("https://github.com/dhnam-data/Datafile/raw/refs/heads/main/StatisticsHealthCare/data/3.%EC%83%81%EA%B4%80%EB%B6%84%EC%84%9D.xlsx")
print(df.shape)
df.head()
```

```python
from scipy import stats

# h20_cin: 경상소득
# h2005_3aq2: 민간의료보험 월납부액
# df = df[['h20_cin', 'h2005_3aq2']].dropna()


correlation = stats.pearsonr(df['h20_cin'], df['h2005_3aq2'])

print(correlation)

display(df.corr(method='pearson'))
```

```python
!pip install pingouin -q
import pingouin as pg

correlation = pg.pairwise_corr(
    data = df,
    columns = ['h20_cin', 'h2005_3aq2'],
    method='pearson'
)

display(correlation)

display(pg.rcorr(df, method='pearson'))
```
