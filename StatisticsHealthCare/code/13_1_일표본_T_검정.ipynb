# 1.일표본 T 검정

귀무가설(H0): 가구의 월평균 민간의료보험료는 29.8 만원이다. [H0: μ1 ＝ 29.8]

대립가설(H1): 가구의 월평균 민간의료보험료는 29.8 만원과 다르다. [H1: μ1 ≠ 29.8]

```python
# 라이브러리 불러오기
import pandas as pd
from scipy import stats
pd.set_option('display.max_columns', None) # 컬럼 전체보기

# 엑셀파일 불러오기 pd.read_excel('파일경로')
df = pd.read_excel("https://github.com/dhnam-data/Datafile/raw/refs/heads/main/StatisticsHealthCare/data/1.%ED%8F%89%EA%B7%A0%EB%B9%84%EA%B5%90.xlsx")

df.head()
df.columns


# 'h2005_3aq2(민간의료보험월납입액)'의 결측치 삭제
a = df['h2005_3aq2'].dropna()

# 기술통계
display(a.describe())

# mu0 값 설정
mu = 29.8

# alternative 기본값: 'two-sided', 단측검정: a < mu = 'less', a > mu = 'greater'
stats_result = stats.ttest_1samp(a, popmean=mu, alternative='two-sided')
print(stats_result)
```
