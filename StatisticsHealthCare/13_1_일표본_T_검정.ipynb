귀무가설(H0): 조사대상자의 혈중 중성지방 수치는 200mg/dL와 같다. [H0: μ1 ＝ 200]

대립가설(H1): 조사대상자의 혈중 중성지방 수치는 200mg/dL와 같지 않다. [H1: μ1 ≠ 200]

```python
# 라이브러리 불러오기
import pandas as pd
from scipy import stats

# 엑셀파일 불러오기 pd.read_excel('파일경로')
df = pd.read_excel("https://github.com/dhnam-data/Datafile/raw/refs/heads/main/StatisticsHealthCare/data/SampleData%204.xlsx")
pd.set_option('display.max_columns', None) # 컬럼 전체보기

# 'HE_TG'의 결측치 삭제
a = df['HE_TG'].dropna()

# 기술통계
display(a.describe())

# mu0 값 설정
mu = 200

# alternative 기본값: 'two-sided', 단측검정: a < mu = 'less', a > mu = 'greater'
stats_result = stats.ttest_1samp(a, popmean=mu, alternative='two-sided')
print(stats_result)

shapiro_result = stats.shapiro(a)
print(shapiro_result)
```
